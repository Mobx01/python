# colored scatterplots

```python
iris = pd.read_csv('iris.csv') #dataset pf a flower iris which have 3 types : setosa,virginica, virgicolor
iris
```

```python
#if i want to see relation between sepal length and petal length
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'])
plt.xlabel('SepalLengthCm')
plt.ylabel('PetalLengthCm')
plt.title('relation between sepal and petal length')
```

```python
# if we want to identify different types of iris on the scatter plot we will give each type a color
# we assign numbers to the tpye : setosa-0, versicolor-1 and virginica-3
iris['Species']  =iris['Species'].replace({'Iris-setosa':0,
                        'Iris-versicolor':1,
                        'Iris-virginica':2})


# now we use a parameter c in the scatter fn
# c=iris['Species'] -> will assign different color to different number in species colmn
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'],c=iris['Species'])
plt.xlabel('SepalLengthCm')
plt.ylabel('PetalLengthCm')
plt.title('relation between sepal and petal length')

# cmap='name of color map' -> modify assigned color to different type.names-summer,winter,jet erc
# plt.colorbar() -> gives the relation berween color and its related values
plt.colorbar()

#aplha - parameter that controllers opacity of graph (less value of alpha less opacity)(value = 0 to 1)
```

# annotation

```python
batter = pd.read_csv('batter.csv')
sample_df = batter.head(100).sample(25,random_state=5)# from top 100 batsman choose random 25
sample_df
```

```python
#plot a scatter plot avg vs strike_rate with name and dot size varies with runs he scored
plt.figure(figsize=(25,7))
plt.scatter(sample_df['avg'],sample_df['strike_rate'],s=sample_df['runs'])
plt.xlabel('avg')
plt.ylabel('strike rate')

# anotation -> we can place a text on graph
#use plt.text(xcordinate,ycordinate,'label at that point')

for i in range(sample_df.shape[0]) :
    plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i],sample_df['batter'].values[i])
```

# placing horizontal and vertical lines on graph

```python
#plt.axhline(y value) ->places a horizontal line at y = value
#plt.axvline(x value) ->places a vertical line at x = value
# can plot any number of horizontal or vertical lines

plt.figure(figsize=(15,7))
plt.scatter(sample_df['avg'],sample_df['strike_rate'],s=sample_df['runs'])
plt.xlabel('avg')
plt.ylabel('strike rate')

plt.axhline(130,color='red') 
plt.axvline(30,color='red') 

for i in range(sample_df.shape[0]) :
    plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i],sample_df['batter'].values[i])
```

# subplots
- different way to plot graphs
- can be used to plot multiple graphs at once

```python
#plotting using subplots


# syntax -  here fig-> figure object and ax-> axis object
# fig ,ax =plt.subplot()
# ax.<graph_type(graph_name)>(x-axis row , y-axis row)

fig,ax = plt.subplots()
ax.scatter(batter['avg'],batter['strike_rate']) #this plots the graph

#for adding title  - ax.set_title('title of graph')
#for adding x axis label - ax.set_xlabel('label of x axis')
#for adding y axis label - ax.set_ylabel('label of y axis')
#for changing size we can pass a parameter figsize with  value in a tuple to the function subplot
#   - fig,ax = plt.subplot(figsize=(x size , y size))
# for printing the graph in terminal enviroment - fig.show()
#for chaning color marker etc we do it same as in normal plot fn by passing parameter with its value

ax.set_title("avg vs strike rate")
ax.set_xlabel('avg')
ax.set_ylabel('Strike rate')
```

Behind the scene when we call the subplot fn
plot.subplot() , it returns 2 value 1st a figure object which is the graph where plotting is done 
2nd is the axis object which represents the axis of the graph(x and y axis)

plotting multiple graphs together using subplots

```python
# we pass 2 parametres nrows and ncolumns with there respective values
# fig,ax= plt.subplot(nrows= number of rows , ncols = number of columns )

# Here the ax we contain array of aixs of graphs
#now for printing we will access the element of axis array by indexing
# for first graph ax[0] for 2nd ax[1]
#when there is both more than 1 columns and row ax will be 2d array 
#for accesing first graph of first row ax[0][0] or ax[0,0] and so on


# we can share value of x axis (means it will only show on lower graph and upper grpahs x value will be hidden)
# -> pass a paramtere sharex=True in the subplot function
# fig,ax= plt.subplot(nrows= number of rows , ncols = number of columns ,sharex=True)

#just like x we can share y axis values(means only one graph will hawve value for 2 grpahs which are presented adjcently)
# -> pass a paramtere sharey=True in the subplot function
# fig,ax= plt.subplot(nrows= number of rows , ncols = number of columns ,sharey=True)
```

```python
#plot 2 graphs side by side - run vs avg and strike_rate vs avg

fig , ax = plt.subplots(nrows=2,ncols=1,sharex=True) #-> makes a figure with two graphs in 2 rows and 1 column format

ax[0].scatter(batter['avg'],batter['strike_rate'] ,color='black') #ploting the first graph in the figure
ax[0].set_title("avg vs strike rate")
ax[0].set_xlabel('avg')
ax[0].set_ylabel('Strike rate')



ax[1].scatter(batter['avg'],batter['runs'])#ploting the 2nd graph
ax[1].set_title("avg vs runs")
ax[1].set_xlabel('avg')
ax[1].set_ylabel('runs')

```

```python
#plot 4 graphs where top 2 are run vs avg and strike_rate vs avg and bottom 2 are histogram -  strike rate and avg

fig , ax = plt.subplots(nrows=2,ncols=2,figsize=(15,12)) #-> makes a figure with two graphs in 2 rows and 1 column format

ax[0][0].scatter(batter['avg'],batter['strike_rate'] ,color='black') #ploting the first graph of first row in the figure
ax[0][0].set_title("avg vs strike rate")
ax[0][0].set_xlabel('avg')
ax[0][0].set_ylabel('Strike rate')

ax[0][1].scatter(batter['avg'],batter['runs'])#ploting the 2nd graph of 1st row
ax[0][1].set_title("avg vs runs")
ax[0][1].set_xlabel('avg')
ax[0][1].set_ylabel('runs')

ax[1][0].hist(batter['strike_rate'] ,color='black') #ploting the first graph of 2nd row in the figure
ax[1][0].set_title("number of players in a certain strike rate range")
ax[1][0].set_xlabel('strike rate')
ax[1][0].set_ylabel('nmber of players')

ax[1][1].hist(batter['avg'] ,color='red') #ploting the 2nd graph of 2nd row in the figure
ax[1][1].set_title("number of players in a certain avg range")
ax[1][1].set_xlabel('avg')
ax[1][1].set_ylabel('nmber of players')


```

# one more way to plot multiple graphs together 

```python
# fig = plt.figure() -> creates a figure
#now we can add as many axises we want 
# ax1 = fig.add_subplot(<no of rows>,<no of cols>,<this graphs location>)
# ax1.<graph_name>(x-axis rows,y-axis row,hidden parameters(no neccesaary))

#ax2 = ... and so on we can plot as many graphs together

#graph number moves through rows like 
# 1 ,2 (if 2 cols)
# 3, 4 ...
```

```python
fig = plt.figure()

ax1 = fig.add_subplot(1,2,1)
ax1.scatter(batter['avg'],batter['strike_rate'] ,color='black')
ax1.set_title("avg vs strike rate")
ax1.set_xlabel('avg')
ax1.set_ylabel('Strike rate')

ax2 = fig.add_subplot(1,2,2)
ax2.scatter(batter['avg'],batter['runs'] ,color='black')
ax2.set_title("avg vs run")
ax2.set_xlabel('avg')
ax2.set_ylabel('run')
```

# 3d surface plots

```python
# plot graph of x^2 + y^2

x = np.linspace(-10,10,100) #create a array of 100 values between -10 and 10
y = np.linspace(-10,10,100)

xx,yy = np.meshgrid(x,y)

z = xx**2 + yy**2

fig = plt.figure(figsize=(15,10))
ax = plt.subplot(projection='3d')
ax.plot_surface(xx,yy,z)
```

```python
# can change color map using cmap='<color map name>'

# for getting color details we can store the graph in a variable then use colorbar function and send the variable as parameter

fig = plt.figure(figsize=(15,10))
ax = plt.subplot(projection='3d')
p = ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

```python
z = np.sin(xx)+np.cos(yy)
fig = plt.figure(figsize=(15,10))
ax = plt.subplot(projection='3d')
p = ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

# contour plot

- represents 3d graph in 2d 
- image of 3d graph from top view -> lines help to understand depth (observe thier color)

```python
fig = plt.figure(figsize=(15,10))
ax = plt.subplot(projection='3d')
p = ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)

#plot this graph in 2d
fig = plt.figure(figsize=(15,10))
ax = plt.subplot()
p = ax.contour(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

contourf plot
- another way to represent 3d graph into 2d
- filled contour graph -> color is filled in the contour graph to get better visulaization 

```python
fig = plt.figure(figsize=(15,10))
ax = plt.subplot()
p = ax.contourf(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

# Heatmap

```python
#construct a grid where y axis has overs(1,2,...,20) and x axis will have all the 6 balls of over.
#make the for each ball of each over print total number of sixes(combined) hitted on that ball in ipl history


delivery=pd.read_csv('IPL_Ball_by_Ball_2008_2022.csv')

#currently some over has ball up to 10 but we will take out only for first 6 balls
temp_df = delivery[(delivery['ballnumber'].isin(([1,2,3,4,5,6])) & (delivery['batsman_run']== 6) )]
temp_df

#now will constuct a pivot table 

grid = temp_df.pivot_table(index='overs',columns='ballnumber',values='batsman_run',aggfunc='count')

#now plot a heatmap on this pivot
plt.imshow(grid)
plt.colorbar()
```
