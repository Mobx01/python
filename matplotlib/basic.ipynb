# labeling


```python
#plt.title("title of graph") -> gives title to the grpah
#plt.xlabel("lable of x axis") -> names the x axis 
#plt.ylabel("lable of y axis") -> names the y axis 

plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma']) 

plt.title("R Sharma vs V kohli")
plt.xlabel("Year")
plt.ylabel("Run Scored")
```

# color ,line(widht,style) and marker(size) 

```python

#plt.plot(x-axis,y-axis,color="name/hexcode of color") ->changes color of line

#plt.plot(x-axis,y-axis,linestyle='type') -> changes line style
 #    type - dashed -> gives dashed lines
 #         - dotted -> forms doted lines
 #         - dashdot-> consecutive dash and dots

#plt.plot(x-axis,y-axis,linewidth=<value in int>) -> change widht of line


#plt.plot(x-axis,y-axis,marker='type of marker',markersize=<size in int>) -> add markers to the points for which data is provided
#                  type - D -> diamond
#                       - + -> +
#                       - . -> .
#                       - > -> arrow
#                       - < -> arrow
#                       - o -> bigger dot


#plt.plot(x-axis,y-axis,label="name of the line") -> name of the line (this only assigns name to line )
#but to show the label on graph we use a function legend
#plt.legend(loc='lacation at which label is situated') -> default is 'best'- sys figures out where to place the label 
plt.plot(batsman['index'],batsman['V Kohli'],color='black',linestyle='dotted',linewidth=2,marker='D',label='V Kohli')# giving color using name
plt.plot(batsman['index'],batsman['RG Sharma'],color='#D9F10F',linestyle='dashed',linewidth=4,marker='+',markersize=10,label='R Sharma') #giving color using hexcode

plt.title("R Sharma vs V kohli")
plt.xlabel("Year")
plt.ylabel("Run Scored")
plt.legend()
```

# limiting axes

```python
#when there is one value which is compartevly very large than other values(in this casee 400000)
# other changes in value will look plan in the graph . so we can set a value to any axis so that graph doesnt look planar because of a very bug value
# this is called trimming

#plt.ylim(start,end) -> set limit to the y axis that only values from staring to ending will be showen
#plt.xlim(start,end) -> set limit to the x axis that only values from staring to ending will be showen
 

price = [48000,50000,52000,50500,53400,52000,400000] 
year =[2015,2016,2017,2018,2019,2020,2021]

plt.plot(year,price)
plt.ylim(0,100000);
plt.xlim(2017,2020);
```

# scatter plots

 - uses: finding correlations
 
 - 2d plot is special case of scatter plot.
 - discrete points are plotted

```python
#plt.scatter(x_axis,y_axis) -> plots a scattered graph 

x = np.linspace(-10,10,50) # array of 50 numbers equally distint . between -10 to 10
y = 10*x + 3 + np.random.randint(0,300,50) #array where y[i] = x[i]*10 + 3 + random num(between 0 to 300)


plt.scatter(x,y)
plt.grid()
```

multiple dataset on a single graph

```python
df = pd.read_csv('batsman_season_record.csv')
df
```

```python
# if we try to use multiple plot fc call -> won't work as expected because
# graph will be overwritted so no benifit because data is lost.eg-
plt.bar(df['batsman'],df['2015'])
plt.bar(df['batsman'],df['2016'])
plt.bar(df['batsman'],df['2017'])

plt.title("No of fours each season ")
plt.xlabel("season")
plt.ylabel("no of fours")

# in the below generated graph we can see blue one(represents 2015 data) is lost for most of the players

```

```python
# there is no direct fn to get multiple data side by side. we will use
# some jugad for that in matplotlb
# we will use width parameter for getting  different data side by side 


# we replace name of batsaman by number in xaxis and perform action to get data side by side

plt.bar(np.arange(df.shape[0])-0.3,df['2015'],width= .3,label='2015')
plt.bar(np.arange(df.shape[0]),df['2016'],width= .3,label='2016')
plt.bar(np.arange(df.shape[0])+0.3,df['2017'],width= .3,label='2017')

plt.title("No of fours each season ")
plt.xlabel("season")
plt.ylabel("no of fours")
plt.legend()


#then we will get our names back using plt.xticks() functions
#plt.xticks(current name , new name)
plt.xticks(np.arange(df.shape[0]) , df['batsman'])

plt.show()
```

```python
#label overlaping - problem
#happens when name on x axis are large
#soln- use of parameter rotation in  xticks
children = [20,43,12,23,34] #np of children who loves certain color
colors = ['red(and a very big name causing overlaping)','blue(very very big name soooo big)','green','yellow','pink'] #colors

plt.bar(colors,children)
plt.xticks(rotation='vertical')

#alternative- plt.xticks(rotation=<degree>)-> use degree 90 to get vertical
```

stacked bar chart


```python
#for ploting one graph on top pf of other we use a parameter bottom on the bar fn and its value is the data below it

plt.bar(df['batsman'],df['2015'],label='2015')
plt.bar(df['batsman'],df['2016'],bottom=df['2015'],label='2016')
plt.bar(df['batsman'],df['2017'],bottom=df['2015']+df['2016'],label='2017')

plt.title("No of fours each season ")
plt.xlabel("season")
plt.ylabel("no of fours")
plt.legend()
```

# histogram
- divided numerical data into group
- univarient analysis
- use case -> frequency count

```python
data = [32,45,56,10,15,27,61]

#plt.hist(data) -> creayes a histogram
plt.hist(data) # autpmayicallly creates bins if not provided
#bins - slots(range) of data
# for manully providing bins -> parameter named bins is usec
# bins =(10,25,40,55.70)-> creates bin of range:(10-25),(25-40),(40-55),(55-70)
plt.hist(data,bins=(10,25,40,55.70))
```

# piechart
- univariate or bivariate analysis
- categorical vs numerical
- use case -> to find the contibution on a standard scale

```python
data=[23,45,100,20,40]
plt.pie(data)

#plt.pie(data of pie)-> generates pie chart
```

```python
#plot with labels
# pass a array of labels as a value to parameter labels
data=[23,45,100,20,40]
subjects =['maths','science','english','sst','cs']
plt.pie(data,labels=subjects)
```

```python
df = pd.read_csv('gayle-175.csv') #-> data of the match in which gayle scored 175.(contributions of different batsman on that match)
df

plt.pie(df['batsman_runs'],labels=df['batsman']) #-> represents contribution of each batsman in teams total
```

```python
# autopct = '%0.1f%%' - passing this as a parametershows percentage of contribution
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%')

# can set colors manually by passing a paratmeter color having value as array of colors
```

```python
# extracting a pie
# explode=[array having valie representing amount in which the pie should explode]
#  value - 0 means original (0 explosion)

plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%',explode=[0,0,0,0,0.1,0])

#shadow-parameter which give shadow to a pie chrt
```
