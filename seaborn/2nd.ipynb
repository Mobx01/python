## Figure level fn -> catplot

- strip plot 
     - plots scatter plot for categorical data
     - one of the column must be of numerical data
     


```python
sns.stripplot(data=tips,x='day',y='total_bill')
```

```python
#for removing the noise to make the graph in a single strip we pass a parameter jitter=False
sns.stripplot(tips,x='day',y='total_bill',jitter=False)

#jitter is the noise added by seaborn to improve visibility of each point
```

```python
# we can also controll the amount of jitter in the graph by passing value to the jitter parameter
sns.stripplot(tips,x='day',y='total_bill',jitter=0.3)
```

```python
#using catplot
#-> figure level fn
sns.catplot(tips,x='day',y='total_bill',kind='strip')
```

```python
#hue can also be used to differentitate between values of a column
sns.stripplot(tips,x='day',y='total_bill',hue='sex')
```

- swarmplot
     - similar to strip but in swarm behind the scene ther runs a algo which makes distributionn of data more visiblly noticable.
     - lke where the data is more dense the graph whill have larger branches there

```python
#using axis level fn
sns.swarmplot(tips,x='day',y='total_bill')


#using figure level fn
sns.catplot(tips,x='day',y='total_bill',kind='swarm')

#can use hue similar to strip
sns.catplot(tips,x='day',y='total_bill',kind='swarm',hue='sex')
```

- box plot <br>
A boxplot is a standardized way of displaying the distribution of data based on a five number summary (“minimum”, first quartile [Q1], median, third quartile [Q3] and “maximum”). It can tell you about your outliers and what their values are. Boxplots can also tell you if your data is symmetrical, how tightly your data is grouped and if and how your data is skewed.



```python
#plotted wrt a categorical data

sns.boxplot(tips,x='sex',y='total_bill') #with categorical data sex
sns.boxplot(tips,x='day',y='total_bill') # with categorical data day
```

```python
#using figure level fn
sns.catplot(tips,x='day',y='total_bill',kind='box') 
```

```python
# if we use hue it will make different box of each value of column passed to the hue parameter

#plot box plot for total_bill on each day for all gender
sns.boxplot(tips,x='day',y='total_bill',hue='sex')
```

```python
# single boxplot -> onyl numerical value
# no categorical col
# just don't use  x parameter

sns.boxplot(tips,y='total_bill')
```

- violinplot
    - boxplot+kdeplot
    - in the middle of each structure of the plot there is a boxplot


```python
sns.violinplot(tips,x='day',y='total_bill')
```

```python
#using catplot
sns.catplot(tips,x='day',y='total_bill',kind='violin')
```

```python
# hue - will get plots same as in boxplot
sns.violinplot(tips,x='day',y='total_bill',hue='sex')
```

```python
#split- if the column passed to hue has only 2 value we can plot the too plots in same structue
#       by passing paramter split=True

sns.violinplot(tips,x='day',y='total_bill',hue='sex',split=True)
```

- Barplot
     - estimation of values 
     - calculate some aggerated of a categorical value with numerical value
     - When there are multiple observations in each category, it also uses bootstrapping to compute a confidence interval around the estimate, which is plotted using error bars

```python
sns.barplot(tips,x='sex',y='total_bill')
#plots average value of total_bill for male and female
# the black lines are error bars
```

```python
#hue - plots side by side bar for different value of the column passed to the hue
sns.barplot(tips,x='sex',y='total_bill',hue='smoker')
```

```python
# for using another aggregate fn 
#pass a parameter estimator=estimate fn
import numpy as np
sns.barplot(tips,x='sex',y='total_bill',estimator=np.std) # now plots on basis of standard deviatin
```

- point plot <br>
        -same as barplot just replace the bar with a point and connect it

```python
sns.pointplot(tips,x='sex',y='total_bill')
```

```python
#with hue

sns.pointplot(tips,x='sex',y='total_bill',hue='smoker')
```

- countplot
     - A special case for the bar plot is when you want to show the number of observations in each category rather than computing a statistic for a second variable. This is similar to a histogram over a categorical, rather than quantitative, variable

```python
sns.countplot(tips,x='sex') #reflects the count of each gender eho visited 
```

```python
# hue - same , plots side by side
sns.countplot(tips,x='sex',hue='day')
```

```python
# faceting 
# can be done in catplot only

sns.catplot(tips,x='sex',y='total_bill',col='smoker',row='time')
# colwrap works same 
```

- Regplot
     - plots a scatter plot + tries to plot a linear regression line for that data
     - axis level fn
both functions(reg and lmplot) draw a scatterplot of two variables, x and y, and then fit the regression model y ~ x and plot the resulting regression line and a 95% confidence interval for that regression.


```python

sns.regplot(tips,x='total_bill',y='tip')

# here the line is the linear regression line on data total_bill vs tip
# the lighter shade of blue line  is the error 
```

- lmplot
      - same as relplot
      - figure level fn
      - we can use hue parameter here

```python
sns.lmplot(tips,x='total_bill',y='tip')

# plottoing for different gender using hue
sns.lmplot(tips,x='total_bill',y='tip',hue='sex')
```

- residplot
     - residual plot
     - plots the distance of all the points from linear regression line
     - represents error of prediction


```python
sns.residplot(tips,x='total_bill',y='tip')
```

- facetgrid <br>
      - second way to plot  facet plot
      - More customizations

```python
# syntax
# g = sns.FacetGrid(data=<dataset name>,,col='<name of colm for which facetwill be plted in coln>',row='<con for which facet will be plted in row>') -> this creates and object of facet grid
# make a empty graph object.
# g.map(<graph type>,<x axis> , <y axis>)

g = sns.FacetGrid(tips,col='day',row='time')
g.map(sns.residplot,'tip','total_bill')
```

- pairplot 
       - pairwise relationship
       - automatically detects numerical column in dataset and plots a scatter splot betwwen different pair of columns
       - plots histogram for pair where x and y are same pair

```python
iris
```

```python
sns.pairplot(iris)
```

```python
# with hue - replaces histogram with kde plot
sns.pairplot(iris,hue='species')
```

- pairgrid
      - same as pair plpt with more features
      - benift : can plot any type of graph

```python
# g = sns.PairGrid(dataset)
# g.map(<map>) 

g = sns.PairGrid(iris)
g.map(sns.scatterplot)


# with hue
g1 = sns.PairGrid(iris,hue='species')
g1.map(sns.scatterplot)
```

```python
# we can choose different graph for diagonal and non diagonal
# g=sns.PairGrid(dataset)
# g.map_diag(<map>)
# g.map_offdiag(<map>)

g = sns.PairGrid(iris)
g.map_offdiag(sns.scatterplot)
g.map_diag(sns.histplot)
```

```python
# we can plot different graph of upper part of diagonal and lower part of diagona
#g.map_upper(<mapname>)
#g.map_lower(<mapname>)


g = sns.PairGrid(iris)
g.map_offdiag(sns.scatterplot)
g.map_diag(sns.histplot)
g.map_upper(sns.kdeplot)
```

```python
# we can control which columns we have to work on
#vars=['column1',...]

g = sns.PairGrid(iris,vars=['sepal_width','petal_width'])
g.map_offdiag(sns.scatterplot)
g.map_diag(sns.histplot)
g.map_upper(sns.kdeplot)

```

- Joint plot
     - scatter plot between numerical column and histogram for both columns

```python
# sns.jointplot(data='dataset',x='x_col',y='y_col')
sns.jointplot(data=tips,x='total_bill',y='tip')
```

```python
# when we use kind parameter we can change graph type which is plotted
#kind='graph name'
#graph name = scatter = scatter +histogram (default)
#             kde = 2D kde + 1D kde
#             hist = 2d hostogram + 1d histogram
#             reg = regression plot + 1d (hist + kde)

sns.jointplot(data=tips,x='total_bill',y='tip',kind='kde')
```

- Jointgrid
     - same as joint plot with extra features

```python
# g = sns.JointGrid(data=dataset,x='x_col',y='y_col')
# g.plot(first graph name , seconf graph name)


g = sns.JointGrid(tips,x='total_bill',y='tip')
g.plot(sns.scatterplot,sns.histplot)
```
